```python
import matplotlib.pyplot as plt
import numpy as np
from open_ephys.analysis import Session
import os
import re
from scipy.signal import butter, filtfilt, lfilter
from collections import defaultdict
import scipy.stats as stats

# ===============================
# CONFIGURATION & SESSION SETUP
# ===============================
# Here we need to load both recordnodes:
#  - Record Node 106: Online Filtered
#  - Record Node 111: Offline Raw (for filtering)
# The following user setting is only used to select the main offline node for sync and ADC1:
w_offline = 1  # use Record Node 111 for sync, ADC1, and offline methods

directory = 'SAP-55_2025-06-09_11-12-19_001'
session = Session(directory)
print('Pass Initial Test, Moving on to Loading Session...\n')

def scan_experiment_structure(session_dir, record_node_name="Record Node 106"):
    node_path = os.path.join(session_dir, record_node_name)
    if not os.path.exists(node_path):
        print(f"Record Node directory not found: {node_path}")
        return {}
    experiment_info = defaultdict(list)
    for item in sorted(os.listdir(node_path)):
        exp_path = os.path.join(node_path, item)
        if os.path.isdir(exp_path) and item.startswith("experiment"):
            recordings = [r for r in sorted(os.listdir(exp_path))
                          if os.path.isdir(os.path.join(exp_path, r)) and r.startswith("recording")]
            experiment_info[item] = recordings
    return experiment_info

# Scan offline structure (you may also scan online if desired)
record_node_name_offline = "Record Node 111"
structure = scan_experiment_structure(directory, record_node_name_offline)

# Flatten experiment-recording pairs to loop over
flat_recordings = []
for exp_name, rec_list in structure.items():
    for rec in rec_list:
        flat_recordings.append((exp_name, rec))

# ===============================
# FILTER FUNCTIONS & PROCESSING METHODS
# ===============================
def bandpass_filter_common(data, fs, lowcut=100, highcut=1000, order=4, method='filtfilt'):
    nyq = 0.5 * fs
    b, a = butter(order, [lowcut/nyq, highcut/nyq], btype='band')
    if method == 'lfilter':
        return lfilter(b, a, data)
    else:  # default: 'filtfilt'
        return filtfilt(b, a, data)

# Processing method functions:
def process_online_filtered(data_tuple, fs):
    # For online: use Record Node 106 data only (first element of tuple)
    # Here, the online processing does no additional filtering.
    data_online = data_tuple[0]
    # Assume channels: emg1_raw = data[:, 2], emg2_raw = data[:, 3]
    # and the “online” output is simply the second channel (CH 11)
    return data_online[:, 3]

def process_offline_forward(data_tuple, fs):
    # Use offline data only (second element of tuple) and apply forward filtering via lfilter.
    data_offline = data_tuple[1]
    emg1 = data_offline[:, 2]
    emg2 = data_offline[:, 3]
    diff = emg1 - emg2
    filtered = bandpass_filter_common(diff, fs, method='lfilter')
    return filtered

def process_offline_2way(data_tuple, fs):
    # Use offline data and apply bidirectional filtering (filtfilt).
    data_offline = data_tuple[1]
    emg1 = data_offline[:, 2]
    emg2 = data_offline[:, 3]
    diff = emg1 - emg2
    filtered = bandpass_filter_common(diff, fs, method='filtfilt')
    return filtered

def process_offline_2way_abs(data_tuple, fs):
    # As above but take absolute value.
    return np.abs(process_offline_2way(data_tuple, fs))

# Function to select processing method based on a string key.
def process_emg_method(method, data_tuple, fs):
    if method == 'Online Filtered Differential':
        return process_online_filtered(data_tuple, fs)
    elif method == 'Offline Forward Filtered':
        return process_offline_forward(data_tuple, fs)
    elif method == 'Offline 2-way Filtered':
        return process_offline_2way(data_tuple, fs)
    elif method == 'Offline 2-way Filtered + Abs':
        return process_offline_2way_abs(data_tuple, fs)
    else:
        raise ValueError("Unknown processing method")

# List of method names (and desired ordering)
processing_methods = [
    'Online Filtered Differential',
    'Offline Forward Filtered',
    'Offline 2-way Filtered',
    'Offline 2-way Filtered + Abs'
]

# ===============================
# MAIN LOOP: ITERATE OVER RECORDINGS
# ===============================
# We loop over recordings as before.
for x, (exp, yy) in enumerate(flat_recordings):
    y = f"recording{x + 1}"
    # For folder navigation, v is determined from record node name:
    v_online = "Record Node 106"
    v_offline = "Record Node 111"

    # Load both recordings (if available)
    try:
        recording_online = session.recordnodes[0].recordings[x]  # online node 106
    except IndexError:
        print(f"No recording {x+1} found in {exp}/{yy} for online node")
        continue
    try:
        recording_offline = session.recordnodes[w_offline].recordings[x]  # offline node 111
    except IndexError:
        print(f"No recording {x+1} found in {exp}/{yy} for offline node")
        continue

    print(f"\nProcessing: x={x}, y={y}, exp={exp}, yy={yy}")

    # ---- Sync Setup (use offline recording for sync and ADC1) ----
    print(f"\n{exp.upper()}, Recording {x+1}")
    print(recording_offline)
    # Check sync lines and add if necessary
    if not any(sync.get("main", False) for sync in recording_offline.sync_lines):
        recording_offline.add_sync_line(1, 100, 'Rhythm Data', main=True)
    recording_offline.compute_global_timestamps()

    metadata = recording_offline.continuous[0].metadata
    channel_names = metadata['channel_names']
    sample_rate = metadata['sample_rate']
    timestamps = recording_offline.continuous[0].timestamps
    data_offline = recording_offline.continuous[0].get_samples(0, len(timestamps))
    # Also load online data:
    timestamps_online = recording_online.continuous[0].timestamps
    data_online = recording_online.continuous[0].get_samples(0, len(timestamps_online))
    # Bundle the two data sets into a tuple (online, offline)
    data_tuple = (data_online, data_offline)

    # ---- ADC1 & Sync Events (from offline) ----
    adc1 = data_offline[:, 4]  # ADC1 channel (adjust if needed)
    events = recording_offline.events
    sync_events = events[(events.line == 1) &
                         (events.processor_id == 100) &
                         (events.stream_name == 'Rhythm Data') &
                         (events.state == 1)]
    sync_timestamps = sync_events['timestamp'].to_numpy()

    # ---- Load MessageCenter ----
    messagecenter_dir = os.path.join(directory, v_offline, exp, yy, "events", "MessageCenter")
    if not os.path.exists(messagecenter_dir):
        print(f"MessageCenter directory not found for {exp}/{yy}")
        continue
    texts = np.load(os.path.join(messagecenter_dir, "text.npy"), allow_pickle=True)
    timestamps_msg = np.load(os.path.join(messagecenter_dir, "timestamps.npy"))
    decoded_texts = [t.decode('utf-8') if isinstance(t, bytes) else str(t) for t in texts]
    message_entries = list(zip(timestamps_msg, decoded_texts))
    print(f"Loaded {len(decoded_texts)} MessageCenter entries")
    for text, tmsg in zip(decoded_texts, timestamps_msg):
        print(f"[Time: {tmsg:.6f} s] Message: {text}")

    # ---- Full Trace Plot (for reference) ----
    # For full trace we show method 'Offline 2-way Filtered + Abs'
    differential_emg_full = process_offline_2way_abs(data_tuple, sample_rate)
    plt.figure(figsize=(15, 4))
    plt.plot(timestamps, differential_emg_full, label="Filtered EMG1 - EMG2", color='purple')
    plt.axhline(0, color='black', linestyle='--', linewidth=0.5)
    plt.title(f"{directory}, {exp}, {yy}, Absolute Value Filtered Differential EMG Signal (EMG1 - EMG2)")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude (μV)")
    plt.grid(True)
    plt.legend()
    plt.ylim(top=15000)
    plt.tight_layout()
    plt.show()

    # ---- Peri-Stimulus Analysis (common indices based on offline ADC1) ----
    pre_event_time = 5    # ms before stim start
    post_event_time = 20  # ms after stim start
    stim_threshold = 0.1  # threshold for ADC1

    # Compute indices for each stim event (using offline timestamps & ADC1)
    peri_indices = []       # will store tuples (idx_pre, idx_post, stim_start_time, stim_end_time)
    segment_stim_start_times = []
    segment_stim_end_times = []
    for stamp in sync_timestamps:
        idx_search_start = np.searchsorted(timestamps, stamp - 0.010)
        idx_search_end   = np.searchsorted(timestamps, stamp + 0.010)
        if idx_search_start < 0 or idx_search_end > len(timestamps):
            continue
        adc1_segment = adc1[idx_search_start:idx_search_end]
        above_idxs = np.where(adc1_segment > stim_threshold)[0]
        if len(above_idxs) == 0:
            continue
        stim_start_idx = idx_search_start + above_idxs[0]
        stim_start_time = timestamps[stim_start_idx]
        idx_pre  = int(stim_start_idx - (pre_event_time/1000)*sample_rate)
        idx_post = int(stim_start_idx + (post_event_time/1000)*sample_rate)
        if idx_pre < 0 or idx_post > len(timestamps):
            continue
        # Determine stim_end_time from ADC1 falling below threshold after stim start
        time_window = timestamps[idx_pre:idx_post] - stim_start_time
        adc1_window  = adc1[idx_pre:idx_post]
        post_onset_mask = (time_window >= 0)
        adc1_post = adc1_window[post_onset_mask]
        time_post = time_window[post_onset_mask]
        below_idxs = np.where(adc1_post < stim_threshold)[0]
        stim_end_time = time_post[below_idxs[0]]*1000 if len(below_idxs) > 0 else post_event_time
        peri_indices.append( (idx_pre, idx_post, stim_start_time, stim_end_time) )
        segment_stim_start_times.append(stim_start_time)
        segment_stim_end_times.append(stim_end_time)

    # ---- For each stimulation amplitude group determined by MessageCenter ----
    # (Grouping code remains the same)
    trigger_pattern = re.compile(r"RHDCONTROL TRIGGER")
    stim_pattern = re.compile(r"starting(?: at)? (\d+\.?\d*)\s*mA", re.IGNORECASE)
    stim_groups = []
    current_group = []
    current_amp = None

    for t, msg in message_entries:
        if stim_pattern.search(msg):
            if current_amp is not None and current_group:
                stim_groups.append((current_amp, current_group))
            current_amp = float(stim_pattern.search(msg).group(1))
            current_group = []
        elif trigger_pattern.search(msg):
            if current_amp is not None:
                current_group.append(t)
    if current_amp is not None and current_group:
        stim_groups.append((current_amp, current_group))

    # ---- For plotting peri-stimulus responses we will iterate over each stim group.
    stim_intensities = []
    m_wave_peaks = []
    h_wave_peaks = []

    for stim_amp, trigger_times_group in stim_groups:
        group_data = []  # will store segments for this group (per event)
        group_start_times = []
        group_end_times = []
        for trig_time in trigger_times_group:
            # Match each trigger time to a peri-stim segment based on start times
            if len(segment_stim_start_times) == 0:
                continue
            idx_match = np.argmin(np.abs(np.array(segment_stim_start_times) - trig_time))
            if abs(segment_stim_start_times[idx_match] - trig_time) < 0.1:
                group_data.append(peri_indices[idx_match])
                group_start_times.append(segment_stim_start_times[idx_match])
                group_end_times.append(segment_stim_end_times[idx_match])
        if not group_data:
            continue

        # For each stim group (i.e. one amplitude), now create a single figure that has 4 subplots,
        # each using one of the processing methods.
        fig, axs = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey=True)
        axs = axs.flatten()
        # For a common time vector, extract from the first segment:
        idx_pre, idx_post, _, _ = group_data[0]
        time_window = timestamps[idx_pre:idx_post] - timestamps[idx_pre + int((pre_event_time/1000)*sample_rate)]
        time_ms = time_window * 1000  # in ms

        for j, method in enumerate(processing_methods):
            # Process the entire continuous trace with the chosen method:
            processed_emg = process_emg_method(method, data_tuple, sample_rate)
            # For each valid peri-stim segment, extract the segment from processed_emg:
            segments = []
            for (i_pre, i_post, stim_start, stim_end) in group_data:
                seg = processed_emg[i_pre:i_post].copy()
                # Baseline correct (using first  pre_event portion)
                baseline = np.mean(seg[:int((pre_event_time/1000)*sample_rate)])
                seg = seg - baseline
                segments.append(seg)
            segments = np.array(segments)
            # Compute average across events:
            avg_response = np.mean(segments, axis=0)
            ax = axs[j]
            for seg in segments:
                ax.plot(time_ms, seg, color='red', alpha=0.4)
            ax.plot(time_ms, avg_response, color='black', linewidth=2, label='Average')
            ax.axvline(0, color='red', linestyle='--', label='Stim')
            ax.set_title(method)
            ax.set_xlabel("Time (ms)")
            ax.set_ylabel("Amplitude (μV)")
            ax.grid(True)
            ax.legend()

        fig.suptitle(f"{exp}, {yy} | Peri-Stimulus EMG (Stim Amp: {stim_amp} mA, N={len(group_data)})", fontsize=14)
        plt.tight_layout(rect=[0, 0, 1, 0.93])
        plt.show()

        # For recruitment curves, we use the average response (from, for example, the 4th method) to extract peaks:
        # Here, we use the 4th method: Offline 2-way Filtered + Abs.
        processed_emg_for_recruit = process_offline_2way_abs(data_tuple, sample_rate)
        group_reps = []
        for (i_pre, i_post, stim_start, stim_end) in group_data:
            seg = processed_emg_for_recruit[i_pre:i_post]
            baseline = np.mean(seg[:int((pre_event_time/1000)*sample_rate)])
            seg = seg - baseline
            group_reps.append(seg)
        group_reps = np.array(group_reps)
        avg_resp = np.mean(group_reps, axis=0)
        # Define windows for M-wave and H-wave (in ms)
        m_wave_start, m_wave_end = 2, 5.0
        h_wave_start, h_wave_end = 6, 10
        m_mask = (time_ms >= m_wave_start) & (time_ms <= m_wave_end)
        h_mask = (time_ms >= h_wave_start) & (time_ms <= h_wave_end)
        m_peak = np.max(avg_resp[m_mask])
        h_peak = np.max(avg_resp[h_mask])
        m_wave_peaks.append(m_peak)
        h_wave_peaks.append(h_peak)
        stim_intensities.append(stim_amp)

    # ---- Recruitment Curve Analysis (Box & Whisker) ----
    m_wave_dict = defaultdict(list)
    h_wave_dict = defaultdict(list)
    for amp, m_peak, h_peak in zip(stim_intensities, m_wave_peaks, h_wave_peaks):
        m_wave_dict[amp].append(m_peak)
        h_wave_dict[amp].append(h_peak)
    sorted_amps = sorted(set(stim_intensities))
    m_wave_data = [m_wave_dict[amp] for amp in sorted_amps]
    h_wave_data = [h_wave_dict[amp] for amp in sorted_amps]
    m_means = [np.mean(peaks) for peaks in m_wave_data]
    m_sems  = [stats.sem(peaks) for peaks in m_wave_data]
    h_means = [np.mean(peaks) for peaks in h_wave_data]
    h_sems  = [stats.sem(peaks) for peaks in h_wave_data]

    fig, ax = plt.subplots(figsize=(10, 6))
    positions = np.arange(len(sorted_amps))
    ax.boxplot(m_wave_data, positions=positions - 0.2, widths=0.3,
               patch_artist=True, boxprops=dict(facecolor='lightblue'),
               medianprops=dict(color='black'), labels=[f"{amp:.1f}" for amp in sorted_amps])
    ax.boxplot(h_wave_data, positions=positions + 0.2, widths=0.3,
               patch_artist=True, boxprops=dict(facecolor='lightgreen'),
               medianprops=dict(color='black'))
    ax.errorbar(positions - 0.2, m_means, yerr=m_sems, fmt='o-', color='blue', label='M-Wave Mean ± SEM')
    ax.errorbar(positions + 0.2, h_means, yerr=h_sems, fmt='o-', color='green', label='H-Wave Mean ± SEM')
    ax.set_xticks(positions)
    ax.set_xticklabels([f"{amp:.1f}" for amp in sorted_amps])
    ax.set_xlabel("Stimulation Intensity (mA)")
    ax.set_ylabel("Peak Amplitude (μV)")
    ax.set_title(f"{exp}, {yy}, Recruitment Curves (Box & Whisker)")
    ax.grid(True)
    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from open_ephys.analysis import Session
import os
import re
from scipy.signal import butter, filtfilt
from collections import defaultdict
import scipy.stats as stats


# ==== Select Session ====
w = 1  # recordnode index 1 = Record Node 111 RAW
       # recordnode index 0 = Record Node 106 is Online Filtered


# ==== Define Filters ====
def bandpass_filter(data, fs, lowcut=100, highcut=1000, order=2):
    nyq = 0.5 * fs
    b, a = butter(order, [lowcut / nyq, highcut / nyq], btype='band')
    return filtfilt(b, a, data)

def full_filter(data, fs):
    return bandpass_filter(data, fs)

# ==== Load Session ====
#directory = 'SAP-55_2025-06-09_11-12-19_001'

#directory = 'SNARE-38_2025-07-24_13-43-28_002'
#directory = 'SNARE-38_TEST2025-07-24_13-46-09_003'
#directory = 'SNARE-38-TEST1_2025-07-24_13-24-34_001'
#directory = 'SNARE-38-TEST2_2025-07-24_15-06-10_001'
#directory = 'SNARE-38-TEST2_2025-07-24_15-19-20_001'
#directory = 'SNARE-38-TEST2_2025-07-24_15-59-14_001'
directory = 'SNARE-38-TEST3_2025-07-24_16-05-17_002'

session = Session(directory)
print('Pass Initial Test, Moving on to Loading Session...\n')

# ==== Loop through experiments and recordings ====

for x in range(4):  # assuming up to 4 recordings per experiment
    y = f"recording{x + 1}"
    v = f"Record Node 106" if w == 0 else f"Record Node 111"
    
    #################################################################################
    # BASED ON SESSION STRUCTURE< MAKE CHANGES HERE
    # THIS IS THE CASE IF EXPERIMENT 1 ONLY HAS 2 RECORDINGS (SAP-55_2025-06-09_11-12-19_001)
    # IF EXPERIMENT 1 HAS 4 OR N RECORDINGS, CHANGE THE LOOP VARIABLE X IN THE IF STATMENT TO 4 OR N

    if x>=2: # CHANGE HERE BASED ON N OF RECORDINGS IN EXPERIMENT 1 OBJECT
        exp = "experiment2"
        yy = f"recording{x - 1}"
    else: 
        exp = "experiment1"
        yy = y


    ##################################################################
    
    try:
        recording = session.recordnodes[w].recordings[x]
    except IndexError:
        print(f"No recording {x+1} found in {exp}")
        continue

    # ==== Sync Setup ====
    print(f"\n{exp.upper()}, Recording {x+1}")
    print(recording)

    # Only add sync line if a main one doesn't already exist
    has_main_sync = any(sync.get("main", False) for sync in recording.sync_lines)
    if not has_main_sync:
        recording.add_sync_line(1, 100, 'Rhythm Data', main=True)

    recording.compute_global_timestamps()

    metadata = recording.continuous[0].metadata
    channel_names = metadata['channel_names']
    print("Channels:", channel_names, '\n')

    timestamps = recording.continuous[0].timestamps
    data = recording.continuous[0].get_samples(start_sample_index=0, end_sample_index=len(timestamps))
    sample_rate = metadata['sample_rate']

    # ==== EMG Processing ====
    emg1_raw = data[:, 2]
    emg2_raw = data[:, 3]
    differential = (emg1_raw - emg2_raw)
    differential_filt = full_filter((differential), sample_rate)
    differential_emg = np.abs(differential_filt)

    # ==== Load ADC1 (for stim detection) ====
    adc1 = np.abs(data[:, 4])  # adjust if your ADC1 channel index differs

    # ==== Sync Events ====
    events = recording.events
    sync_events = events[(events.line == 1) & (events.processor_id == 100) &
                            (events.stream_name == 'Rhythm Data') & (events.state == 1)]
    sync_timestamps = sync_events['timestamp'].to_numpy()

    # ==== Load MessageCenter ====
    messagecenter_dir = os.path.join(directory, v, exp, yy, "events", "MessageCenter")
    if not os.path.exists(messagecenter_dir):
        print(f"MessageCenter directory not found for {exp}/{yy}")
        continue

    texts = np.load(os.path.join(messagecenter_dir, "text.npy"), allow_pickle=True)
    timestamps_msg = np.load(os.path.join(messagecenter_dir, "timestamps.npy"))
    decoded_texts = [t.decode('utf-8') if isinstance(t, bytes) else str(t) for t in texts]
    message_entries = list(zip(timestamps_msg, decoded_texts))
    print(f"Loaded {len(decoded_texts)} MessageCenter entries")

    # ==== Debug Print ====
    for text, time in zip(decoded_texts, timestamps_msg):
        print(f"[Time: {time:.6f} s] Message: {text}")

    # ==== Full Trace Plot ====
    plt.figure(figsize=(15, 4))
    plt.plot(timestamps, differential_emg, label="Filtered EMG1 - EMG2", color='purple')
    plt.axhline(0, color='black', linestyle='--', linewidth=0.5)
    plt.title(f"{directory}, {exp}, {yy}, Absolute Value Filtered Differential EMG Signal (EMG1 - EMG2)")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude (μV)")
    plt.grid(True)
    plt.legend()
    plt.ylim(top=15000)
    plt.tight_layout()
    plt.show()

    # ==== Peri-Stimulus Analysis with Shaded Stimulus Duration ====
    pre_event_time = 5   # ms before calculated stim start
    post_event_time = 20 # ms after calculated stim start
    stim_threshold = 0.1 # ADC1 threshold to detect pulse (adjust if needed)

    peri_stim_segments = []
    segment_stim_end_times = []
    segment_stim_start_times = []

    for stamp in sync_timestamps:
        # First, find approximate sample index around the sync pulse
        idx_center = np.searchsorted(timestamps, stamp)
        # Define a generous window to locate true stim-onset in ADC1
        idx_search_start = np.searchsorted(timestamps, stamp - 0.010)  # 10 ms before sync
        idx_search_end   = np.searchsorted(timestamps, stamp + 0.010)  # 10 ms after sync

        if idx_search_start < 0 or idx_search_end > len(timestamps):
            continue

        adc1_segment_for_detection = adc1[idx_search_start:idx_search_end]
        time_segment_for_detection = timestamps[idx_search_start:idx_search_end]

        # Detect true stim start: first crossing above threshold in that window
        above_idxs = np.where(adc1_segment_for_detection > stim_threshold)[0]
        if len(above_idxs) == 0:
            continue
        stim_start_idx = idx_search_start + above_idxs[0]
        stim_start_time = timestamps[stim_start_idx]

        # Now define peri-stimulus window around this stim_start_time
        idx_pre  = int(stim_start_idx - (pre_event_time / 1000) * sample_rate)
        idx_post = int(stim_start_idx + (post_event_time / 1000) * sample_rate)

        if idx_pre < 0 or idx_post > len(timestamps):
            continue

        # Extract windows
        time_window = timestamps[idx_pre:idx_post]
        time_zeroed = time_window - stim_start_time  # in seconds

        emg_window  = differential_emg[idx_pre:idx_post]
        adc1_window = adc1[idx_pre:idx_post]

        # Detect stim end: first time after stim_start where ADC1 falls back below threshold
        post_onset_mask = (time_zeroed >= 0)
        adc1_post_onset = adc1_window[post_onset_mask]
        time_post_onset = time_zeroed[post_onset_mask]

        below_idxs = np.where(adc1_post_onset < stim_threshold)[0]
        if len(below_idxs) > 0:
            stim_end_time = time_post_onset[below_idxs[0]] * 1000  # convert to ms
        else:
            stim_end_time = post_event_time

        # Save for grouping
        peri_stim_segments.append((time_zeroed * 1000, emg_window, stim_end_time))
        segment_stim_start_times.append(stim_start_time)
        segment_stim_end_times.append(stim_end_time)

    # Helper to find most recent stim amplitude message
    def find_latest_amplitude_msg(before_time):
        pattern = re.compile(r"starting(?: at)? (\d+\.?\d*)\s*mA", re.IGNORECASE)
        for t, msg in reversed(message_entries):
            if t < before_time:
                m = pattern.search(msg)
                if m:
                    return m.group(1)  # just number as string
        return None

    group_size = 5
    n_groups = len(peri_stim_segments) // group_size

    # Lists to accumulate recruitment curve data per recording
    stim_intensities = []
    m_wave_peaks = []
    h_wave_peaks = []

    for i in range(n_groups):
        group_data = peri_stim_segments[i*group_size:(i+1)*group_size]
        group_start_times = segment_stim_start_times[i*group_size:(i+1)*group_size]
        group_end_times   = segment_stim_end_times[i*group_size:(i+1)*group_size]

        # Stack to compute average EMG trace
        time_ms_matrix = np.stack([gd[0] for gd in group_data])  # shape: (5, window_samples)
        emg_matrix     = np.stack([gd[1] for gd in group_data])

        avg_time_ms = time_ms_matrix[0]  # identical for all in group
        avg_emg     = np.mean(emg_matrix, axis=0)
        avg_end_ms  = np.mean(group_end_times)

        stim_amp_str = find_latest_amplitude_msg(group_start_times[0])
        if stim_amp_str is not None:
            try:
                stim_amp_float = float(stim_amp_str)
            except:
                stim_amp_float = None
        else:
            stim_amp_float = None

        # Skip group if stim amplitude not found or invalid
        if stim_amp_float is None:
            stim_amp_float = 0.0  # or skip plotting recruitment point?

        stim_intensities.append(stim_amp_float)

        plt.figure(figsize=(7, 4))
        # Plot individual EMG traces
        for (_, trace, _) in group_data:
            plt.plot(avg_time_ms, trace, color='red', alpha=0.6)
        # Plot average EMG
        plt.plot(avg_time_ms, avg_emg, color='black', linewidth=2, label='Average EMG')

        # Vertical lines for stim start (0) and stim end (avg_end_ms), excluded from legend
        plt.axvline(x=0, color='red', linestyle='--', label='Stim Start/End')
        #plt.axvline(x=1.0, color='red', linestyle='--', label='_nolegend_')
        
        #plt.axvspan(0, 1.0, color='red', alpha=0.3)

        #---------------Here we can manually set the red box width for the stimm ----------------#
        plt.axvline(x=avg_end_ms, color='red', linestyle='--', label='_nolegend_')
        plt.axvspan(0, avg_end_ms, color='red', alpha=0.3)

        # Define M-Wave and H-Wave windows in ms
        m_wave_start, m_wave_end = 2, 5.0
        h_wave_start, h_wave_end = 6, 10

        # Shaded regions
        plt.axvspan(m_wave_start, m_wave_end, color='blue', alpha=0.3)
        plt.axvspan(h_wave_start, h_wave_end, color='green', alpha=0.3)

        # Find peak in M-Wave window
        m_wave_mask = (avg_time_ms >= m_wave_start) & (avg_time_ms <= m_wave_end)
        m_wave_time_window = avg_time_ms[m_wave_mask]
        m_wave_emg_window = avg_emg[m_wave_mask]

        m_wave_peak_idx = np.argmax(m_wave_emg_window)
        m_wave_peak_time = m_wave_time_window[m_wave_peak_idx]
        m_wave_peak_amp = m_wave_emg_window[m_wave_peak_idx]
        m_wave_peaks.append(m_wave_peak_amp)  # accumulate for recruitment curve

        # Find peak in H-Wave window
        h_wave_mask = (avg_time_ms >= h_wave_start) & (avg_time_ms <= h_wave_end)
        h_wave_time_window = avg_time_ms[h_wave_mask]
        h_wave_emg_window = avg_emg[h_wave_mask]

        h_wave_peak_idx = np.argmax(h_wave_emg_window)
        h_wave_peak_time = h_wave_time_window[h_wave_peak_idx]
        h_wave_peak_amp = h_wave_emg_window[h_wave_peak_idx]
        h_wave_peaks.append(h_wave_peak_amp)  # accumulate for recruitment curve

        # Plot vertical lines at peak positions with label
        plt.axvline(x=m_wave_peak_time, color='blue', linestyle=':', linewidth=2,
                    label=f'M-Wave Peak: {m_wave_peak_amp:.1f} μV at {m_wave_peak_time:.2f} ms')
        plt.axvline(x=h_wave_peak_time, color='green', linestyle=':', linewidth=2,
                    label=f'H-Wave Peak: {h_wave_peak_amp:.1f} μV at {h_wave_peak_time:.2f} ms')

        # Annotate peak amplitude text near the vertical line (offset a bit)
        plt.text(m_wave_peak_time, m_wave_peak_amp + 200, f'{m_wave_peak_amp:.1f} μV',
                color='blue', fontsize=9, ha='center')
        plt.text(h_wave_peak_time, h_wave_peak_amp + 200, f'{h_wave_peak_amp:.1f} μV',
                color='green', fontsize=9, ha='center')

        plt.title(
            f"{exp}, {yy}, Peri-Stimulus EMG "
            f"(Events {i*group_size+1}-{(i+1)*group_size}) | Stim Amp: {stim_amp_float} mA"
        )
        plt.xlabel("Time (ms)")
        plt.ylabel("Amplitude (μV)")
        plt.grid(True)
        plt.legend()
        plt.ylim(top=15000)

        # X-ticks at each integer ms
        min_ms = int(np.floor(avg_time_ms[0]))
        max_ms = int(np.ceil(avg_time_ms[-1]))
        plt.xticks(np.arange(min_ms, max_ms + 1, 1))

        plt.tight_layout()
        plt.show()

    # ==== Group peak amplitudes by stimulation intensity ====
    m_wave_dict = defaultdict(list)
    h_wave_dict = defaultdict(list)

    for amp, m_peak, h_peak in zip(stim_intensities, m_wave_peaks, h_wave_peaks):
        m_wave_dict[amp].append(m_peak)
        h_wave_dict[amp].append(h_peak)

    # Sort by stimulation intensity
    sorted_amps = sorted(set(stim_intensities))

    m_wave_data = [m_wave_dict[amp] for amp in sorted_amps]
    h_wave_data = [h_wave_dict[amp] for amp in sorted_amps]

    # Compute mean and SEM for overlay
    m_means = [np.mean(peaks) for peaks in m_wave_data]
    m_sems  = [stats.sem(peaks) for peaks in m_wave_data]
    h_means = [np.mean(peaks) for peaks in h_wave_data]
    h_sems  = [stats.sem(peaks) for peaks in h_wave_data]

    # ==== Plot Recruitment Curves as Box-and-Whisker ====
    fig, ax = plt.subplots(figsize=(10, 6))

    # Boxplots
    positions = np.arange(len(sorted_amps))  # for spacing
    print(f"len(sorted_amps) = {len(sorted_amps)}")
    print(f"len(m_wave_data) = {len(m_wave_data)}")
    for i, peaks in enumerate(m_wave_data):
        print(f"  Intensity {sorted_amps[i] if i < len(sorted_amps) else '?'}: {len(peaks)} peaks")
    box1 = ax.boxplot(m_wave_data, positions=positions - 0.2, widths=0.3,
                    patch_artist=True, boxprops=dict(facecolor='lightblue'),
                    medianprops=dict(color='black'), labels=[f"{amp:.1f}" for amp in sorted_amps])
    box2 = ax.boxplot(h_wave_data, positions=positions + 0.2, widths=0.3,
                    patch_artist=True, boxprops=dict(facecolor='lightgreen'),
                    medianprops=dict(color='black'))

    # Overlay mean ± SEM
    ax.errorbar(positions - 0.2, m_means, yerr=m_sems, fmt='o-', color='blue', label='M-Wave Mean ± SEM')
    ax.errorbar(positions + 0.2, h_means, yerr=h_sems, fmt='o-', color='green', label='H-Wave Mean ± SEM')

    # Aesthetics
    ax.set_xticks(positions)
    ax.set_xticklabels([f"{amp:.1f}" for amp in sorted_amps])
    ax.set_xlabel("Stimulation Intensity (mA)")
    ax.set_ylabel("Peak Amplitude (μV)")
    ax.set_title(f"{exp}, {yy}, Recruitment Curves (Box & Whisker)")
    ax.grid(True)
    ax.legend()
    plt.tight_layout()
    plt.show()
```
